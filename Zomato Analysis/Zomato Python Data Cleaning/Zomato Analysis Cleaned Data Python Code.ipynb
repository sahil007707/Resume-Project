## Data Cleaning & Preprocessing

```python
df.drop(columns=["url", "phone", "address", "reviews_list","menu_item"], inplace=True, errors="ignore")
```

```python
df.head()
```

```python
df=df[df["rate"]!="NEW"] # Dropping the "NEW" values from the column 
```

```python
df["rate"] = df["rate"].str.replace(" ", "")   # remove spaces
df["rate"] = df["rate"].str.replace("/5", "")  # remove '/5' for clean visualization
```

```python
df["rate"] = pd.to_numeric(df["rate"], errors="coerce") # Converting into a numeric column
```

```python
df.isna().sum() # Missing or null values
```

```python
df.dropna(subset=["location"], inplace=True)
df['rate'] = df['rate'].fillna(df['rate'].median())
df['rest_type'] = df['rest_type'].fillna('Unknown')
df['dish_liked'] = df['dish_liked'].fillna('Not Available')
df = df.dropna(subset=['cuisines'])
```

```python
import warnings
warnings.filterwarnings("ignore")
```

```python
# Remove commas and spaces, then convert to numeric 
df["approx_cost(for two people)"] = (
    df["approx_cost(for two people)"]
    .str.replace(",", "")   # remove commas
    .str.strip()            # remove leading/trailing spaces
)

# Convert to numeric
df["approx_cost(for two people)"] = pd.to_numeric(
    df["approx_cost(for two people)"], errors="coerce"
)
```

```python
df['approx_cost(for two people)'] = (
    df['approx_cost(for two people)']
    .fillna(df['approx_cost(for two people)'].median())
)
```
